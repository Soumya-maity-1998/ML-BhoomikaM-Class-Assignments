```python
import numpy as np
```

#### For particle detector, in general the statistics is Poissonian. The standard deviation of a Poisson distribution is the square root of the mean. The standard error is the standard deviation divided by the square root of the number of measurements.


```python
N = 100  # Number of counts in 1 hour. i.e. the mean of the Poisson distribution.
sigma = np.sqrt(N) # Standard deviation of the Poisson distribution.
stderr = sigma/(N) # Standard error of the mean.
```

```python
print(stderr)
print(stderr*100, '%') # Standard error of the mean in percentage.
```

## Second Part

```python
Std_err = 1 # in percentage. This is the standard error of the mean that we want to achieve.
# 1/np.sqrt(N) * 100 = 1. we need to solve for N. 
N = (100/Std_err)**2  
print('Total counts needed to achieve a standard error of the mean of 1%:', N)
```
