```python
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd 
```

```python
# deformation parameter - beta2 
# Energy - E 
beta2 = np.array([0.10, 0.20, 0.30, 0.40, 0.50])
Energy = np.array([0.90, 0.80, 0.60, 0.40, 0.30])
```

```python
beta2.mean()
```

```python
mean_beta2 = beta2.mean()
mean_Energy = Energy.mean() 

print("The mean of coefficient of deformation and energy are: ", mean_beta2, mean_Energy)
```

```python
beta2.var()
```

```python
var_beta2 = beta2.var()
var_Energy = Energy.var()

print("The Variances are ", var_beta2, 'and', var_Energy, 'respectively.' )
```

```python
covariance_matrix = np.cov(beta2, Energy)
```

```python
print(covariance_matrix)
```

```python
df = pd.DataFrame({'beta2': beta2, 'Energy': Energy})
```

```python
df
```

```python
df.info()
```

```python
mean_beta2 = df['beta2'].mean()
mean_Energy = df['Energy'].mean()
beta2_var = df['beta2'].var()
Energy_var = df['Energy'].var()
```

```python
print("The mean of coefficient of deformation and energy are: ", mean_beta2, 'and ', mean_Energy)
```

```python
print('The corresponding variances are: ', beta2_var, 'and', Energy_var)
```

```python
cov = df['beta2'].cov(df['Energy'])
```

```python
print('The covariance is ', cov)
```

```python
corr = df['beta2'].corr(df['Energy'])
```

```python
print('The correlation is', corr)
```
