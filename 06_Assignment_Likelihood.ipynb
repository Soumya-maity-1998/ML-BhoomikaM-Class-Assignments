```python
import numpy as np 
```

```python
# Given Informations:-

sigma = 15.0    # Standard Deviation. (in KeV) 
Energies = np.array([5470, 5495, 5482, 5505, 5478])     # in KeV 

# Likelihood for a single data point

def single_likelihood(y, theta, sigma):
    factor = 1.0/(np.sqrt(2.0*np.pi*sigma**2))
    exponent = np.exp((-(y - theta)**2)/(2.0*sigma**2)) 
    return factor*exponent 

# Joint likelihood for all data points 

def joint_likelihood(data, theta, sigma):
    likelihood = 1.0 
    for y in data:
        likelihood = likelihood*single_likelihood(y, theta, sigma) 
    return likelihood 

# Search for Maximum Likelihood Estimate (MLE) 

thetas = np.linspace(5450, 5520, 250)
likelihoods = []
for theta in thetas: 
    L = joint_likelihood(Energies, theta, sigma) 
    likelihoods.append(L) 

likelihoods = np.array(likelihoods) 

best_theta = thetas[np.argmax(likelihoods)] # armax returns the index where the array is maximum

print('Peak Energy is', best_theta, 'KeV')
print('\nMaximum Joint Likelihood', np.max(likelihoods))
```
